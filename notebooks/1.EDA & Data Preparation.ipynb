# Univariate Analysis of categorical variables

```python
plt.figure(figsize=(16, 15))
for i, feature in enumerate(categorical_columns, 1):
    plt.subplot(2, 2, i)
    sns.countplot(data=train_df, x=feature, color="#3DBDD7")
    plt.title(f'Distribution of {feature}')
    plt.xlabel(feature)
    plt.ylabel('Count')
    plt.xticks(rotation=45)

plt.tight_layout()
plt.suptitle("Categorical Feature Distributions", fontsize=16, y=1.02, fontweight='bold')
plt.show()
```

# Univariate Analysis of numerical variables

```python
import math

# split columns by unique value count
discrete_columns = [
    column for column in numerical_columns if train_df[column].nunique() <= 10
]
continuous_columns = [
    column for column in numerical_columns if column not in discrete_columns
]


# function to plot grid
def plot_grid(column_list, plot_type="hist"):
    rows = math.ceil(len(column_list) / 3)
    plt.figure(figsize=(15, 3.5 * rows))

    for i, feature in enumerate(column_list, 1):
        plt.subplot(rows, 3, i)
        if plot_type == "hist":
            plt.suptitle("Numerical Feature (Continuous Column) Distributions", fontsize=16, y=1.02, fontweight='bold')
            sns.histplot(
                data=train_df, x=feature, kde=True, color="#FF569C", bins=30
            )
        else:
            plt.suptitle("Numerical Feature (Discrete Column) Distributions", fontsize=16, y=1.02, fontweight='bold')
            sns.countplot(data=train_df, x=feature, color="#FF569C")
        plt.title(f"Distribution of {feature}", fontweight="bold")

    plt.tight_layout()
    plt.show()


# plotting continuous columns
plot_grid(continuous_columns, plot_type="hist")

# plotting discrete columns
plot_grid(discrete_columns, plot_type="count")
```

# Bivariate Analysis of categorical variables

```python
plt.figure(figsize=(16, 15))
for i, feature in enumerate(categorical_columns, 1):
    plt.subplot(2, 2, i)
    sns.countplot(data=train_df, x=feature, hue='satisfaction', palette='mako')
    plt.title(f'Distribution of {feature} by Satisfaction')
    plt.xlabel(feature)
    plt.ylabel('Count')
    plt.xticks(rotation=45)

plt.tight_layout()
plt.suptitle("Categorical Feature Distributions by Satisfaction", fontsize=16, y=1.02, fontweight='bold')
plt.show()
```

# Bivariate Analysis of numerical variables

```python
def bivariate_plot(column_list, is_continuous=True):
    rows = math.ceil(len(column_list) / 3)
    plt.figure(figsize=(16, 3.8 * rows))

    for i, feature in enumerate(column_list, 1):
        plt.subplot(rows, 3, i)

        if is_continuous:
            sns.histplot(data=train_df, x=feature, hue="satisfaction", kde=True, element="step", palette="crest", bins=30,)
        else:
            sns.countplot(data=train_df, x=feature, hue="satisfaction", palette="crest")

        plt.title(f"{feature} by Satisfaction", fontweight="bold")
        plt.xlabel("")
        plt.ylabel("Count")

    plot_type_title = "Continuous" if is_continuous else "Discrete"
    plt.suptitle(f"Numerical Feature ({plot_type_title}) Distributions by Satisfaction", fontsize=16, y=1.01, fontweight="bold",)
    plt.tight_layout()
    plt.show()

bivariate_plot(continuous_columns, is_continuous=True)
bivariate_plot(discrete_columns, is_continuous=False)
```

# Data Cleaning

```python
df = train_df.copy()
```

```python
# Filled missing values with median(Robust to outliers)
df['Arrival Delay in Minutes'] = df['Arrival Delay in Minutes'].fillna(df['Arrival Delay in Minutes'].median())
```

# Feature Engineering

```python
df['Total_Delay'] = (df['Departure Delay in Minutes'] + df['Arrival Delay in Minutes'])
```

```python
df['Delayed_Flight'] = (df['Total_Delay'] > 15).astype(int)
```

```python
service_columns = [
    'Inflight wifi service', 'Ease of Online booking', 'Gate location',
    'Food and drink', 'Online boarding', 'Seat comfort', 'Inflight entertainment',
    'On-board service', 'Leg room service', 'Baggage handling', 
    'Checkin service', 'Inflight service', 'Cleanliness'
]

df['Average_Service_Rating'] = df[service_columns].mean(axis=1)
```

**A long-haul flight is generally defined as any flight typically covering a distance of over 3,000 to 4,000 miles**

```python
df['Long_Haul'] = (df['Flight Distance'] > 3000).astype(int)
```
